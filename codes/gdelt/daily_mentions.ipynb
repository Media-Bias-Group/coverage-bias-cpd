```python
def concatenate_daily_mentions(datestring, input_dir, output_path):
    
    zip_files = sorted(input_dir.glob(f"{datestring}*.mentions.CSV.zip"))

    daily_frames = []

    for zpath in zip_files:
        try:
            with zipfile.ZipFile(zpath) as z:
                csv_name = z.namelist()[0]

                with z.open(csv_name) as f:
                    df = pl.read_csv(
                        f,
                        separator="\t",
                        has_header=False,
                        ignore_errors=True
                    )

                    daily_frames.append(df)

        except Exception as e:
            print(f"Error processing {zpath}: {e}")

    daily_mentions = pl.concat(daily_frames, how="vertical")

    daily_mentions.write_parquet(output_path)
```

```python
start = date(2017, 1, 1)
end = date(2017, 12, 31)

base = "http://data.gdeltproject.org/events/"
current = start

while current <= end:
    time.sleep(0.2)
    datestr = current.strftime("%Y%m%d")

    concatenate_daily_mentions(datestr, Path("D:\\gdelt_data\\2017_mentions"), Path(r"D:\gdelt_data\2017_mentions_output") / f"{datestr}.mentions.parquet")
    print(f"Processed {datestr}")
    current += timedelta(days=1)
```
